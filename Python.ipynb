```python
#What is the output of the following Python program?
a = [10, 20, 30]
b = a
c = a.copy()

b[0] = 100
c[1] = 200

print(a)
print(b)
print(c)
print(a is b)
print(a is c)
```

```python
a = [[1, 2], [3, 4]]
b = a.copy()
c = a[:]

a[0].append(99)
b[1] = [7, 8]
c[0] = [10, 20]

print(a)
print(b)
print(c)
print(a[0] is b[0])
print(a[1] is b[1])
```

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(10))
print(add_item(20))
print(add_item(30, []))
print(add_item(40))
```

What is the difference between shallow copying and deep copying?

```python
x = 10
result = [x * y for x in range(3) if x % 2 == 0 for y in range(2)]

print(result)
print(x)
```

```python
def create_functions():
    functions = []

    for i in range(3):
        functions.append(lambda: i * 2)

    return functions

funcs = create_functions()

print([f() for f in funcs])
```

```python
import copy

original = {
    "user": "Ravi",
    "scores": [80, 90]
}

shallow = copy.copy(original)
deep = copy.deepcopy(original)

original["scores"].append(100)
shallow["user"] = "Asha"
deep["scores"].append(70)

print(original)
print(shallow)
print(deep)
```

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])

b = a[:, 1:]
c = a[:, 1:].copy()

b[0, 0] = 99
c[1, 0] = 88

print(a)
print(b)
print(c)
print(a.shape)
print(a[a > 4])
```

How does Python manage memory? Explain reference counting and garbage collection at a high level.

```python
import pandas as pd

employees = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Asha", "Ravi", "Mina"]
})

sales = pd.DataFrame({
    "id": [1, 1, 2, 4],
    "amount": [100, 200, 150, 300]
})

result = employees.merge(sales, on="id", how="left")

print(result)
print(result["amount"].sum())
print(result["amount"].isna().sum())
```
